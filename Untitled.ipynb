# Nullclines

- Una de las herramientas más útiles para analizar sistemas no lineales de ecuaciones diferenciales es la *nullcline*. 

- Para un sistema con n ecuaciones, xj-nullcline es el conjunto de puntos donde j ecuaciones desaparecen, al establecer dicho conjunto de ecuaciones diferenciales iguales a cero.

- El conjunto de *nullclines* generalmente se separan en regiones en las que los componentes xj del campo vectorial apuntan en la dirección positiva o negativa. 

- Al determinar todas las pendientes cero, esto nos permite descomponernos en una colección de conjuntos abiertos, en cada uno de los cuales el campo vectorial apunta en una "cierta dirección".

- Las intersecciones de las líneas nulas x e y producen los puntos de equilibrio.
- Esto es más fácil de entender con un ejemplo:

*x' = alpha x - beta x y*

*y' = gamma x y - delta y*

```python
import numpy as np
import pylab as p
import scipy.integrate as integrate

def dr(r, f):
    return alpha*r - beta*r*f

def df(r, f):
    return gamma*r*f - delta*f

def derivs(state, t):
    """
    Map the state variable [rabbits, foxes] to the derivitives
    [deltar, deltaf] at time t
    """
    #print t, state
    r, f = state # rabbits and foxes
    deltar = dr(r, f) # change in rabbits
    deltaf = df(r, f) # change in foxes
    return deltar, deltaf

alpha, delta = 1, .25
beta, gamma = .2, .05

# the initial population of rabbits and foxes
r0 = 20
f0 = 1

t = np.arange(0.0, 100, 0.1)

y0 = [r0, f0] # the initial [rabbits, foxes] state vector
y = integrate.odeint(derivs, y0, t)
r = y[:,0] # extract the rabbits vector
f = y[:,1] # extract the foxes vector

p.figure()
p.plot(t, r, label='rabbits (x)')
p.plot(t, f, label='foxes (y)')
p.xlabel('time (years)')
p.ylabel('population')
p.title('Population trajectories (Lotka-Volterra model)')
p.grid()
p.legend()
#p.savefig('lotka_volterra.png', dpi=150)


p.figure()
p.plot(r, f,'-')
p.xlabel('rabbits')
p.ylabel('foxes')
p.title('phase plane')


# make a direction field plot with quiver
rmax = 1.1 * r.max()
fmax = 1.1 * f.max()
R, F = np.meshgrid(np.arange(-1, rmax), np.arange(-1, fmax))
dR = dr(R, F)
dF = df(R, F)
C = np.hypot(dR,dF)
p.quiver(R, F, dR, dF, C, pivot='tip')


R, F = np.meshgrid(np.arange(-2, rmax, .1), np.arange(-2, fmax, .1))
dR = dr(R, F)
dF = df(R, F)

p.contour(R, F, dR, levels=[0], linewidths=3, colors='red')
p.contour(R, F, dF, levels=[0], linewidths=3, colors='black')
p.ylabel('foxes')
p.plot([0,5],[0,5],'go')
p.plot(21,1,"o")
p.title('trajectory, direction field and nullclines')

#p.savefig('lotka_volterra_pplane.png', dpi=150)


p.show()
```

- En este ejemplo que presenta oscilaciones, las intersecciones de las nullclines son estados estacionarios inestables, por lo que el sentido de hacia donde va el sistema no converge en ninguno de los dos puntos, siempre y cuando las condiciones iniciales no empiecen exactamente en esos puntos.

```python
y0 = [0, 0] # the initial [rabbits, foxes] state vector
y = integrate.odeint(derivs, y0, t)
r = y[:,0] # extract the rabbits vector
f = y[:,1] # extract the foxes vector

p.figure()
p.plot(t, r, label='rabbits')
p.plot(t, f, label='foxes')
p.xlabel('time (years)')
p.ylabel('population')
p.title('population trajectories at populations = 0')
p.grid()
p.legend()

y0 = [5, 5] # the initial [rabbits, foxes] state vector
y = integrate.odeint(derivs, y0, t)
r = y[:,0] # extract the rabbits vector
f = y[:,1] # extract the foxes vector

p.figure()
p.plot(t, r, label='rabbits')
p.plot(t, f, label='foxes')
p.xlabel('time (years)')
p.ylabel('population')
p.title('population trajectories at populations = 5')
p.grid()
p.legend()
p.show()
```

### Otro ejemplo:
*A' = muA + k1 b^10/(b^10 + kD^10) - gammaA a*

*B' = k2 a - gammaB b*

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

def da(a, b):
    return muA + k1*pow(b, 10)/(pow(b,10)+pow(kD,10)) - gammaA*a

def db(a, b):
    return k2*a - gammaB*b

def derivs(state, t):
    """
    Map the state variable [rabbits, foxes] to the derivitives
    [deltar, deltaf] at time t
    """
    #print t, state
    r, f = state # rabbits and foxes
    deltar = da(r, f) # change in rabbits
    deltaf = db(r, f) # change in foxes
    return deltar, deltaf

muA = 0.2
k1 = 1.8
kD = 40
k2= 0.5
gammaA = 0.1
gammaB = 0.1

# the initial population of rabbits and foxes
t = np.arange(0.0, 200, 0.1)

y0 = [10, 10] 
y = integrate.odeint(derivs, y0, t)
r1 = y[:,0] 
f1 = y[:,1] 

fig = plt.figure(tight_layout=True)
gs = gridspec.GridSpec(2, 1,hspace=0)

ax = fig.add_subplot(gs[0,0])
ax.plot(t, r1, label='A')
ax.plot(t, f1, label='B')
ax.set_xlabel('time (minutes)')
ax.set_ylabel('Concentration')
ax.set_title('Trajectories (Positive Feedback)',fontsize=12)
ax.grid()
ax.legend()
#p.savefig('lotka_volterra.png', dpi=150)

y0 = [30, 40] 
y = integrate.odeint(derivs, y0, t)
r2 = y[:,0] 
f2 = y[:,1] 

ax = fig.add_subplot(gs[1, 0])
ax.plot(t, r2, label='A')
ax.plot(t, f2, label='B')
ax.set_xlabel('time (minutes)')
ax.grid()

plt.figure()
plt.plot(r1, f1,'g-')
plt.plot(r2,f2,'g-')
plt.xlabel('[A]')
plt.ylabel('[B]')
# make a direction field plot with quiver
rmax = 31
fmax = 120
R, F = np.meshgrid(np.arange(-1, rmax,4), np.arange(-1, fmax,6))
dA = da(R, F)
dB = db(R, F)
C = np.hypot(dA,dB)
plt.quiver(R, F, dA, dB, C, pivot='tip', width=0.005)


R, F = np.meshgrid(np.arange(-1, rmax, .1), np.arange(-1, fmax, .1))
dA = da(R, F)
dB = db(R, F)

plt.contour(R, F, dA, levels=[0], linewidths=2, colors='red')
plt.contour(R, F, dB, levels=[0], linewidths=2, colors='black')
plt.plot([2,20],[10,100],'bo')
plt.plot(7.5,36.5,"o",color="purple")
plt.title('Nullclines',fontsize=12)

plt.show()
```
